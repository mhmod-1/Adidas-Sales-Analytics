# Outlier Detection

```python
import matplotlib.pyplot as plt

df.boxplot(column=["Price per Unit", "Units Sold", "Total Sales", "Operating Profit"])
plt.show()
```

```python
columns = ["Price per Unit", "Units Sold", "Total Sales", "Operating Profit"]

for col in columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print(f"{col}: {len(outliers)} outliers")
```

# Data Modeling
## Step 1: Preparing the Fact Table

```python
fact_sales = df.copy()

fact_sales.rename(columns={'Retailer ID': 'Retailer_Key'}, inplace=True)

fact_sales.head()
```

## Step 2: Creating Product Dimension

```python
dim_product = fact_sales[['Product']].drop_duplicates().reset_index(drop=True)

dim_product['Product_Key'] = range(1, len(dim_product) + 1)

dim_product = dim_product[['Product_Key', 'Product']]

# Add Product_Key to Fact Table
fact_sales = fact_sales.merge(dim_product, on='Product', how='left')

dim_product.head()
```

```python
df['Product'].nunique()
```

```python
dim_product['Product'].nunique()
```

```python
fact_sales['Product_Key'].isnull().sum()
```

```python
fact_sales.duplicated().sum()
```

## Step 3: Creating Date Dimension

```python
fact_sales['Invoice Date'] = pd.to_datetime(fact_sales['Invoice Date'])

# Create Date Dimension
dim_date = fact_sales[['Invoice Date']].drop_duplicates().sort_values('Invoice Date')

dim_date = dim_date.reset_index(drop=True)

# Create Date Key
dim_date['Date_Key'] = range(1, len(dim_date)+1)

# Date Features
dim_date['Day'] = dim_date['Invoice Date'].dt.day

dim_date['Month'] = dim_date['Invoice Date'].dt.month

dim_date['Month_Name'] = dim_date['Invoice Date'].dt.month_name()

dim_date['Quarter'] = dim_date['Invoice Date'].dt.quarter

dim_date['Year'] = dim_date['Invoice Date'].dt.year

# Arrange columns
dim_date = dim_date[
[
'Date_Key',
'Invoice Date',
'Day',
'Month',
'Month_Name',
'Quarter',
'Year'
]
]

# Add Date Key
fact_sales = fact_sales.merge(
    dim_date[['Invoice Date','Date_Key']],
    on='Invoice Date',
    how='left'
)

dim_date.head()
```

```python
fact_sales['Date_Key'].isnull().sum()
```

```python
dim_date.duplicated().sum()
```

## Step 4: Creating Location Dimension

```python
# Create Location Dimension
dim_location = fact_sales[['Region','State','City']]

dim_location = dim_location.drop_duplicates().reset_index(drop=True)

# Create Location Key
dim_location['Location_Key'] = range(1,len(dim_location)+1)

# Arrange columns
dim_location = dim_location[
[
'Location_Key',
'Region',
'State',
'City'
]
]

# Add Location Key
fact_sales = fact_sales.merge(
    dim_location,
    on=['Region','State','City'],
    how='left'
)

dim_location.head()
```

## 2: Product Analysis

```python
sales_by_product = (
    df.groupby('Product')['Total Sales']
    .sum()
    .sort_values(ascending=False)
)

colors = plt.cm.Blues(
    np.linspace(0.9, 0.3, len(sales_by_product))
)

plt.figure(figsize=(10,5))

sales_by_product.plot(
    kind='bar',
    color=colors
)

plt.title('Total Sales by Product')
plt.xlabel('Product')
plt.ylabel('Total Sales')

plt.xticks(rotation=45)

plt.show()
```

## 3: Regional Analysis

```python
sales_by_region = (
    df.groupby('Region')['Total Sales']
    .sum()
    .sort_values(ascending=False)
)

colors = plt.cm.Blues(
    np.linspace(0.9, 0.3, len(sales_by_region))
)

plt.figure(figsize=(8,5))

sales_by_region.plot(
    kind='bar',
    color=colors
)

plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')

plt.xticks(rotation=0)

plt.show()
```

## 4: Time Analysis

```python
monthly_sales = (
    df.groupby(df['Invoice Date'].dt.to_period('M'))['Total Sales']
    .sum()
)

plt.figure(figsize=(10,5))

monthly_sales.plot(marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')

plt.grid(True)

plt.show()
```

## 5: Sales Method Analysis

```python
sales_method = (
    df.groupby('Sales Method')['Total Sales']
    .sum()
)

colors = [
    '#0D47A1',
    '#1976D2',
    '#42A5F5'
]

plt.figure(figsize=(6,6))

sales_method.plot(
    kind='pie',
    colors=colors,
    autopct='%1.1f%%',
    startangle=90,
    labeldistance=1.1,   
    pctdistance=0.7,    
    wedgeprops={'edgecolor': 'white', 'linewidth': 0.5}
)

plt.title('Sales by Sales Method')
plt.ylabel('')

plt.tight_layout()
plt.show()
```

## Business Insights

- **Men's Street Footwear** generated the highest total sales, making it the best-performing product.

- The **West** region achieved the highest sales among all regions, while the **Midwest** recorded the lowest sales.

- Monthly sales showed a strong upward trend in **2021** compared to **2020**, indicating business growth over time.

- **In-store** was the most effective sales channel, contributing **39.6%** of total sales. **Online** sales had the lowest contribution at **27.5%**.

- The sales distribution suggests that expanding the online sales channel and improving the performance of lower-selling products could further increase overall revenue.


## Conclusion

This analysis provided valuable insights into Adidas sales performance across products, regions, sales methods, and time.

The results showed that Men's Street Footwear was the top-selling product, the West region generated the highest revenue, and In-store was the leading sales channel. In addition, monthly sales demonstrated strong growth throughout 2021 compared to 2020.

Overall, the analysis highlights opportunities to improve online sales, strengthen lower-performing regions, and optimize product strategies to support future business growth.
